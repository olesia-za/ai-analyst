```python
import os
import pandas as pd
from sqlalchemy import create_engine
from dotenv import load_dotenv

load_dotenv()
DATABASE_URL = os.getenv("DATABASE_URL")
```

```python

engine = create_engine(DATABASE_URL)
```

```python
query = """
SELECT *
FROM users
LIMIT 10;
"""

df = pd.read_sql(query, engine)

print(df)
```

```python
query = """
SELECT
    plan,
    COUNT(*) AS users
FROM users
GROUP BY plan
ORDER BY users DESC;
"""

pd.read_sql(query, engine)
```

```python
query = """
SELECT
    ROUND(
        100.0 * AVG(is_certified),
        2
    ) AS certification_rate
FROM enrollments;
"""

pd.read_sql(query, engine)
```
